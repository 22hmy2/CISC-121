```python
# Cyclic rotation problem 1
# Define function with input of array and k, the number of desired rotations
def cyclicrotation(array, k):
    # Create new rotated list
    
    rotated = []

    # Check if array is empty
    
    if array == None:

            return rotated
    else:

    # Find length of array to define index of range where N = maximum index
    
        N = len(array)

    # Loop through each element in array and move to rotated position in rotated list
    
        for i in range (0, N):

        # Ensure the index will be within the range of array 
       
        # if a large number of rotations is given
        
            if (k/N) <= 1:
        
            # number of rotations is within index range, so subtract k directly
            
                rotated.append(array[i - k])
            else:
        
        # number of rotations is outside of index range
        
        # Use floor division to find how many full rotations will occur in k rotations
        
        # and substract the full rotations to k within index range
            
                rotated.append(array[i - (k - ((k // N) * N))])
        
        # Output: new rotated list
    
    return rotated
```

```python
# Frog Jump problem 3

# Define function with inputs x, y, and d where x = position, y = target position, 

# and d = jump distance

def FrogJump(x, y, d):

    # Check if needed # of jumps will bring frog to exact position or past it

    if (y-x) % d == 0:
        
        # if yes, then floor division will return exact number of jumps 
        
        return ((y - x) // d)
    else:
        
        # if not, then floor division will return 1 less jumps necessary

        return (((y - x) // d) + 1)

FrogJump(3, 24, 5)
```
